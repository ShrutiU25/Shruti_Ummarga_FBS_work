## To handle missing value
Delete column - check if important
Delete row - below 10%
Fill the missing value 

```python
employee.drop('BG', axis = 1) #to drop column
```

```python
employee.dropna()
#dropna method will delete rows with null values
```

```python
employee.isna().sum()
```

## 1st way to fill null value

```python
avg = employee['Salary'].mean()
employee.loc[employee['Salary'].isna(), 'Salary'] = avg
```

```python
employee
```

## 2nd way to fill null value

```python
employee['Salary'].fillna(avg, inplace = True)
#automated method to fill null values
```
