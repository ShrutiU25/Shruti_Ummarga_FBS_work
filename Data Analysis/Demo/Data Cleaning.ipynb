## 1st method to fill null values

```python
#data['AirBags'].fillna("Driver only", inplace = True)
#or
data['AirBags'] = data['AirBags'].fillna("Driver only")
```

```python
data.isna().sum()
```

## 2nd option

```python
result = data['AirBags'].mode()
#store mode operation answer
```

```python
data['AirBags'] = data['AirBags'].fillna(result[0])
#make use of answer
```

```python
data.loc[data['Luggage.room'].isna() == True, 'Luggage.room'] = data['Luggage.room'].mean()
```
